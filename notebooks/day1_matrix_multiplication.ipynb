```python
import sys, torch
import numpy as np
import matplotlib.pyplot as plt
print("Python:", sys.executable)
print("Torch version:", torch.__version__)
print("MPS available:", torch.backends.mps.is_available())

# Detect GPU (MPS for Apple Silicon)
device = "mps" if torch.backends.mps.is_available() else "cpu"
print("Using device:", device)
```

```python
# Scalar, Vector, Matrix Basics 

A = torch.randint(0, 10, (3, 3)).to(device)
print(A, A[1,0])
```

```python
# 2. Matrix–Vector Multiplication (Geometric View)

A = torch.tensor([[0,-1],[1,0]]).to(device)
x = torch.tensor([1,2]).to(device)
y = A @ x
print("Original vector x:\n", x)
print("Matrix A:\n", A)
print("Transformed vector :\n", y)


def plot_vectors(vectors, colors, labels, xlim=(-10, 10), ylim=(-10, 10)):
    plt.axhline(0, color='black', linewidth=0.5)
    plt.axvline(0, color='black', linewidth=0.5)

    for v, col, lab in zip(vectors, colors, labels):
        plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color=col, label=lab)

    plt.xlim(xlim)
    plt.ylim(ylim)
    plt.grid(True, linestyle='--', alpha=0.4)
    plt.legend()
    plt.gca().set_aspect('equal')
    plt.show()

def to_numpy(v):
    if isinstance(v, torch.Tensor):
        return v.detach().cpu().numpy()
    return v


plot_vectors(
    vectors=[to_numpy(x), to_numpy(y)],
    colors=["blue", "red"],
    labels=["Original vector x", "Transformed vector A·x"]
)
```

```python
# Stretch matrix

A = torch.tensor([[5,0],[0,1]]).to(device)
x = torch.tensor([1,4]).to(device)
y = A @ x
print("Original vector x:\n", x)
print("Matrix A:\n", A)
print("Transformed vector :\n", y)  

plot_vectors(
    vectors=[to_numpy(x), to_numpy(y)],
    colors=["blue", "red"],
    labels=["Original vector x", "Transformed vector A·x"]
)
```

```python
# Shear matrix

A = torch.tensor([[5,0],[0,1]]).to(device)
x = torch.tensor([1,4]).to(device)
y = A @ x
print("Original vector x:\n", x)
print("Matrix A:\n", A)
print("Transformed vector :\n", y)  

plot_vectors(
    vectors=[to_numpy(x), to_numpy(y)],
    colors=["blue", "red"],
    labels=["Original vector x", "Transformed vector A·x"]
)
```

```python
# Reflection matrix

A = torch.tensor([[1,0],[0,-1]]).to(device)
x = torch.tensor([1,4]).to(device)
y = A @ x
print("Original vector x:\n", x)
print("Matrix A:\n", A)
print("Transformed vector :\n", y)  

plot_vectors(
    vectors=[to_numpy(x), to_numpy(y)],
    colors=["blue", "red"],
    labels=["Original vector x", "Transformed vector A·x"]
)
```
